```python
import pandas as pd
```

```python
data=pd.read_csv('../data/vehicles.csv')
print(data.info())
```

```python
data=data.drop(columns=["id","url","region_url","lat","long","paint_color","image_url","description","VIN","posting_date","region","county"])
```

```python
mask_man = data[["manufacturer", "model"]].dropna()
model2man=dict(zip(mask_man["model"], mask_man["manufacturer"]))
data["manufacturer"]=data["manufacturer"].fillna(data["model"].map(model2man))
data.info()
```

```python
data.duplicated().any()
data=data.drop_duplicates()
data.info()
```

```python
mask_size=data[["size", "model"]].dropna()
size_dict=dict(zip(mask_size["model"], mask_size["size"]))
data["size"]=data["size"].fillna(data["model"].map(size_dict))

data.info()
```

```python
mask_size=data[["model", "drive"]].dropna()
size_dict=dict(zip(mask_size["model"], mask_size["drive"]))
data["drive"]=data["drive"].fillna(data["model"].map(size_dict))

data.info()
```

```python
mask_size=data[["model", "type"]].dropna()
size_dict=dict(zip(mask_size["model"], mask_size["type"]))
data["type"]=data["type"].fillna(data["model"].map(size_dict))

data.info()
```

```python
mask_size = data[["model", "cylinders"]].dropna()
size_dict = dict(zip(mask_size["model"], mask_size["cylinders"]))
data["cylinders"] = data["cylinders"].fillna(data["model"].map(size_dict))

data.info()
```

```python
mask_size = data[["model", "condition"]].dropna()
size_dict = dict(zip(mask_size["model"], mask_size["condition"]))
data["condition"] = data["condition"].fillna(data["model"].map(size_dict))

data.info()
```
